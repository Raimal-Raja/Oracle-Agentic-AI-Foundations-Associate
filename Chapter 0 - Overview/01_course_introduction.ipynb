# What Is Agentic AI?

Traditional AI systems are often designed to respond to a specific input. For example, a user asks a question, and an AI model generates an answer.

**Agentic AI** extends this idea by enabling AI systems to **pursue goals and perform actions**.

Instead of only answering a question, an AI agent can:

* Understand a goal.
* Reason about a task.
* Decide what actions are necessary.
* Select and use external tools.
* Observe the results of its actions.
* Continue working through multiple steps.
* Stop when the task is completed or a stopping condition is reached.

A simplified comparison is:

### Traditional LLM Interaction

```text
User asks a question
        ↓
LLM processes the request
        ↓
LLM generates a response
```

### Agentic AI

```text
User provides a goal
        ↓
Agent understands the goal
        ↓
Agent reasons about the task
        ↓
Agent selects a tool
        ↓
Tool performs an action
        ↓
Agent observes the result
        ↓
Agent reasons about the next step
        ↓
Task completed?
     ↙       ↘
   No         Yes
    ↓          ↓
Continue    Final Response
```

The ability to combine **reasoning, tool usage, and iterative execution** is one of the key characteristics of agentic AI.

# Core Components of an AI Agent

The course introduces three fundamental components of an AI agent:

1. **LLM — Large Language Model**
2. **Tools**
3. **Loop**

These components work together to transform an AI model from a simple text generator into a system capable of performing multi-step tasks.

---

## 1. Large Language Model (LLM)

The **Large Language Model (LLM)** acts as the reasoning and language-processing component of an AI agent.

The LLM can:

* Interpret instructions.
* Understand context.
* Generate responses.
* Analyze information.
* Determine what action may be appropriate next.

However, an LLM by itself has limitations.

For example, it may not have direct access to:

* Real-time external information.
* Enterprise systems.
* Databases.
* APIs.
* Business applications.
* Private organizational data.

This is where **tools** become important.

In an agentic system, the LLM can determine when it needs additional information or when an external action is required.

---

## 2. Tools

**Tools** allow an AI agent to interact with the outside world.

Examples of tools include:

* Web search
* Database queries
* APIs
* Calculators
* File systems
* Python code execution
* Enterprise applications
* Cloud services
* Business systems

For example, if an agent is asked to calculate a complex financial value, it may use a calculator tool.

If it needs customer information, it may query a database.

If it needs real-time information, it may call an external API.

Tools extend the capabilities of the LLM beyond generating text.

---

## 3. The Agent Loop

The **agent loop** describes the iterative process through which an agent works toward a goal.

A simplified agent loop can be represented as:

```text
User Goal
    ↓
Understand the Task
    ↓
Reason About Next Action
    ↓
Select a Tool (if needed)
    ↓
Execute the Tool
    ↓
Observe the Result
    ↓
Reason Again
    ↓
Task Complete?
   ↙     ↘
 No       Yes
 ↓         ↓
Loop     Final Response
```

The loop allows the agent to perform multiple steps rather than generating a single response in one pass.

# Reasoning Patterns

The course introduces reasoning patterns that help explain how agents approach complex tasks.

Two important patterns mentioned are:

* **Chain of Thought**
* **ReAct**

These patterns help us understand how an agent can move from a user request toward a solution through structured reasoning and action.

---

## Chain of Thought

**Chain of Thought (CoT)** is a reasoning approach where a complex problem is approached through a sequence of intermediate reasoning steps.

Conceptually:

```text
Complex Problem
      ↓
Step 1
      ↓
Step 2
      ↓
Step 3
      ↓
Final Answer
```

The important idea is that complex tasks can be solved by decomposing them into smaller steps.

> **Important:** In production applications, developers should generally avoid exposing private internal reasoning traces. Instead, applications should provide concise explanations, results, or relevant intermediate outputs when appropriate.

---

## ReAct: Reasoning + Acting

**ReAct** combines reasoning with actions.

The agent:

1. Reasons about what it should do.
2. Performs an action using a tool.
3. Observes the result.
4. Continues reasoning.
5. Performs another action if necessary.
6. Produces a final answer.

A simplified representation is:

```text
Reason
  ↓
Act
  ↓
Observe
  ↓
Reason Again
  ↓
Act Again (if needed)
  ↓
Final Answer
```

This pattern is particularly useful for agents that need to interact with external tools or systems.
