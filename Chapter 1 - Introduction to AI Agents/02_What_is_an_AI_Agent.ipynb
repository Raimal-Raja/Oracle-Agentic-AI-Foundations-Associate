# Defining AI Agents

## Overview

This lesson introduces one of the most fundamental concepts in agentic AI:

> **What exactly is an AI agent?**

The lesson explains how AI agents differ from:

* Chatbots
* Traditional workflows
* Standalone Large Language Models (LLMs)

The central idea is that an AI agent is not necessarily a new or special AI model. Instead, it is an **architectural pattern** that combines an LLM with tools and an execution loop.

The fundamental model introduced in this lesson is:

```text
LLM + Tools + Loop = LLM-Based AI Agent
```

The lesson also introduces the **agent execution loop**, which allows an agent to:

1. Perceive
2. Reason
3. Act
4. Observe

The agent repeats this cycle until it has enough information to complete the task or reaches a predefined safety limit.

---

# 1. What Is an AI Agent?

Before defining an AI agent, it is useful to compare it with systems that we may already be familiar with.

## Chatbot

A traditional chatbot generally responds to what the user asks.

```text
User
 ↓
Question
 ↓
Chatbot
 ↓
Answer
```

The chatbot primarily focuses on generating a response to the user's input.

---

## Fixed Workflow

A workflow follows a predefined sequence of steps.

For example:

```text
Step 1
  ↓
Step 2
  ↓
Step 3
  ↓
Step 4
  ↓
Complete
```

The sequence is typically defined in advance.

The workflow does not necessarily decide dynamically what should happen next.

---

## AI Agent

An AI agent behaves differently.

Instead of following only a predefined sequence, an agent can dynamically decide what to do next based on:

* The user's goal
* The current state
* Available information
* Tool results
* Previous actions
* The outcome of those actions

A simplified comparison is:

| System         | Behavior                            |
| -------------- | ----------------------------------- |
| Chatbot        | Responds to user questions          |
| Fixed Workflow | Follows a predefined sequence       |
| AI Agent       | Dynamically decides what to do next |

The key shift is:

> **An AI agent dynamically decides what action to take next in pursuit of a goal.**

---

# 2. Classical Definition of an AI Agent

In classical AI, an **agent** can be described as a system that:

1. **Perceives** its environment.
2. **Acts** upon that environment.

A simplified representation is:

```text
Environment
     ↓
   Perceive
     ↓
    Agent
     ↓
     Act
     ↓
Environment Changes
     ↓
   Perceive Again
```

The agent continuously interacts with its environment.

For this course, however, the focus is specifically on:

> **LLM-based AI agents**

These agents use Large Language Models as a core component of their decision-making process.

---

# 3. The Fundamental Model of an LLM-Based Agent

The lesson introduces three fundamental components:

```text
LLM + Tools + Loop
```

Together, these components form the foundation of an LLM-based AI agent.

Let's understand each component.

---

## 3.1 LLM — The Brain

The **Large Language Model (LLM)** can be thought of as the brain of the agent.

It is responsible for:

* Understanding instructions.
* Interpreting context.
* Reasoning about tasks.
* Deciding what should happen next.
* Selecting appropriate actions.

Conceptually:

```text
User Goal
    ↓
LLM
    ↓
Understand + Reason
    ↓
Decide Next Action
```

The LLM provides the intelligence that drives the agent's decision-making.

---

## 3.2 Tools — The Hands

Tools allow an AI agent to interact with the external world.

The lesson uses the analogy:

> **Tools are like the hands of an agent.**

An LLM can reason about what needs to be done, but tools allow the agent to actually perform actions.

Examples of tools include:

* APIs
* Databases
* Code execution
* Web search
* File systems
* External applications
* Cloud services

For example:

```text
LLM decides:
"I need current weather information."

        ↓

Weather Tool

        ↓

Returns current weather

        ↓

LLM uses the result
```

Without tools, the agent may be limited to information available within its context.

---

## 3.3 Loop — The Nervous System

The **loop** keeps the agent's execution process moving.

The lesson compares the loop to a nervous system because it allows the agent to continuously:

* Receive information.
* Reason.
* Take action.
* Observe results.
* Decide what to do next.

The simplified architecture is:

```text
          ┌──────────────┐
          │     LLM      │
          │    Brain     │
          └──────┬───────┘
                 │
                 ↓
          ┌──────────────┐
          │    Tools     │
          │    Hands     │
          └──────┬───────┘
                 │
                 ↓
          ┌──────────────┐
          │     Loop     │
          │ Nervous System│
          └──────┬───────┘
                 │
                 └──────→ Continue Execution
```

Therefore:

```text
LLM = Brain
Tools = Hands
Loop = Nervous System
```

Together, they form the foundation of an LLM-based AI agent.

# 4. Four Key Properties of AI Agents

The lesson identifies four important properties of AI agents.

These are:

1. Goal-directed
2. Autonomous
3. Tool-using
4. Iterative

Let's examine each one.

---

## 4.1 Goal-Directed

An AI agent works toward an objective or goal.

For example:

> "Book me a flight from Austin to Zurich."

The agent's goal is not simply to respond with text.

The goal is to help accomplish the task.

The agent may need to:

1. Understand the departure city.
2. Understand the destination.
3. Determine the travel date.
4. Search available flights.
5. Analyze the results.
6. Present the available options.

The agent is therefore **goal-directed**.

---

## 4.2 Autonomous

An agent is capable of dynamically choosing its next steps.

For example, after receiving a user request, it may decide:

```text
Do I need a tool?
       ↓
     Yes
       ↓
Which tool?
       ↓
Flight Search Tool
       ↓
Call Tool
       ↓
Observe Result
```

The exact sequence may depend on the task and the information available.

This is different from a fixed workflow where every step is predetermined.

---

## 4.3 Tool-Using

An AI agent does not only generate text.

It can interact with external systems using tools.

Examples include:

```text
AI Agent
   │
   ├── Web Search
   │
   ├── Database
   │
   ├── API
   │
   ├── Code Execution
   │
   ├── File System
   │
   └── Enterprise Application
```

This allows agents to perform actions and retrieve information that may not be available directly from the LLM.

---

## 4.4 Iterative

An AI agent works through an iterative process.

The agent can:

```text
Observe
   ↓
Reason
   ↓
Act
   ↓
Observe Again
   ↓
Reason Again
   ↓
Act Again
```

The agent repeats this process until it determines that the task is complete.

This is known as the:

> **Agent Execution Loop**

# 6. LLM vs. LLM-Based AI Agent

Understanding the difference between a standalone LLM and an LLM-based agent is fundamental to agentic AI.

## Standalone LLM

A standalone LLM is generally:

* Passive
* Stateless by default
* Prompt-driven
* Focused on generating text

The basic interaction is:

```text
Prompt
  ↓
LLM
  ↓
Generated Text
  ↓
Done
```

The LLM does not automatically maintain memory between independent calls unless the application provides previous conversation history or another memory mechanism.

It also cannot inherently perform arbitrary external actions without being connected to tools or an application that executes those actions.

---

## LLM-Based AI Agent

An agent wraps the LLM inside an execution architecture.

The agent can:

* Maintain state across multiple steps.
* Call tools.
* Gather information.
* Perform actions.
* Observe results.
* Adjust its approach.
* Decide when the task is complete.
* Ask for help when necessary.

The architecture can be represented as:

```text
User Goal
    ↓
Agent
    ↓
LLM
    ↓
Reason
    ↓
Choose Action
    ↓
Call Tool
    ↓
Observe Result
    ↓
Reason Again
    ↓
Continue or Finish
```

---

# 7. An AI Agent Is an Architectural Pattern

One of the most important concepts in this lesson is:

> **An AI agent is not necessarily a new model. It is an architectural pattern.**

You do not necessarily need a completely new AI model to build an agent.

Instead, you can take an existing LLM and build an architecture around it.

Conceptually:

```text
LLM
 +
Tools
 +
Execution Loop
 +
State
 +
Instructions
 +
Safety Controls
 =
AI Agent
```

The execution loop is particularly important because it transforms the LLM from a passive responder into an active decision-making component.

The fundamental transformation is:

```text
Passive LLM
     ↓
LLM + Decision Loop
     ↓
Active Agent
```

This is one of the most important ideas to remember when learning agentic AI.

# 8. The Agent Execution Loop

Every agent operates through some form of execution cycle.

The lesson introduces four key stages:

1. **Perceive**
2. **Reason**
3. **Act**
4. **Observe**

The cycle can be represented as:

```text
        ┌──────────────┐
        │   Perceive   │
        └──────┬───────┘
               ↓
        ┌──────────────┐
        │    Reason    │
        └──────┬───────┘
               ↓
        ┌──────────────┐
        │     Act      │
        └──────┬───────┘
               ↓
        ┌──────────────┐
        │   Observe    │
        └──────┬───────┘
               │
               └────────→ Repeat
```

Let's understand each stage.

---

## 8.1 Perceive

The agent receives information.

This may include:

* User input
* Tool results
* Database results
* API responses
* Previous observations
* Other environmental information

Example:

```text
User:
"Book a flight from Austin to Zurich."
```

The agent perceives this input.

---

## 8.2 Reason

The LLM analyzes the current situation and decides what should happen next.

For example:

```text
"I need to find available flights."

What can I use?

I have access to a flight search tool.
```

The agent decides that the flight search tool should be used.

This is the decision-making stage.

> **Note:** The lesson refers to possible "hidden reasoning." In practical AI systems, internal reasoning traces should generally not be exposed directly to users. Applications should instead provide appropriate explanations or summaries when needed.

---

## 8.3 Act

The agent performs an action.

For example, it may call a flight search tool.

The tool may receive parameters such as:

```text
Origin: Austin
Destination: Zurich
Date: [Travel Date]
```

The agent sends these parameters to the tool.

---

## 8.4 Observe

The agent receives and evaluates the result of its action.

For example:

```text
Flight Tool Result:

Flight 1 → $500
Flight 2 → $650
Flight 3 → $800
```

The agent now has new information.

It can reason again based on this observation.

This creates the iterative cycle:

```text
Perceive
   ↓
Reason
   ↓
Act
   ↓
Observe
   ↓
Reason Again
```

---

# 9. Agent Execution Loop and Human Problem Solving

The agent execution loop is similar to how humans often solve problems.

Humans typically:

1. Understand a situation.
2. Think about what to do.
3. Try something.
4. Observe the result.
5. Adjust their approach.
6. Try again if necessary.

For example:

```text
Problem
   ↓
Think
   ↓
Try
   ↓
Check Result
   ↓
Adjust
   ↓
Try Again
```

AI agents use a similar pattern:

```text
Perceive
   ↓
Reason
   ↓
Act
   ↓
Observe
   ↓
Adjust
   ↓
Repeat
```

This iterative behavior is a key characteristic of agentic systems.

---

# 10. When Does the Agent Loop Stop?

An agent cannot continue indefinitely.

The execution loop typically terminates when one of two conditions is reached.

## Condition 1: Task Completed

The agent determines that it has enough information to provide a final answer.

```text
Agent
  ↓
Enough Information?
  ↓
Yes
  ↓
Final Answer
```

---

## Condition 2: Safety Limit Reached

The agent reaches a predefined maximum number of iterations.

For example:

```text
Maximum Iterations = 10

Iteration 1
Iteration 2
Iteration 3
...
Iteration 10
    ↓
Stop Execution
```

This is an important safety mechanism.

Without limits, an agent could potentially enter an infinite loop.

---

# 11. Risks of the Agent Execution Loop

The execution loop gives agents their power, but it also introduces risks.

## 11.1 Cost

Each iteration may involve one or more LLM API calls.

More iterations can mean:

```text
More LLM Calls
    ↓
Higher Token Usage
    ↓
Higher Cost
```

Therefore, agent efficiency is an important consideration.

---

## 11.2 Latency

Each model call and tool call takes time.

Multiple iterations can increase the total response time.

For example:

```text
LLM Call
   ↓
Tool Call
   ↓
LLM Call
   ↓
Tool Call
   ↓
LLM Call
   ↓
Final Response
```

This may be slower than a single LLM call.

---

## 11.3 Tool Errors

Tools can fail.

For example:

* API unavailable
* Invalid parameters
* Database error
* Network failure
* Authentication failure

The agent must be designed to handle these situations.

---

## 11.4 Hallucinated Tool Calls

An LLM may sometimes attempt to call a tool incorrectly or generate invalid tool parameters.

Therefore, applications need appropriate validation and error handling.

---

## 11.5 Infinite Loops

An agent may repeatedly perform actions without reaching a conclusion.

For example:

```text
Reason
 ↓
Act
 ↓
Observe
 ↓
Reason
 ↓
Act
 ↓
Observe
 ↓
Repeat Forever
```

This is why maximum iteration limits and other guardrails are important.

---

# 12. Why Orchestration and Guardrails Matter

Because agents operate dynamically, production systems need mechanisms to control their behavior.

Two important concepts are:

* **Orchestration**
* **Guardrails**

### Orchestration

Orchestration controls how the agent's components work together.

It can manage:

* LLM calls
* Tool calls
* Agent loops
* State
* Multi-step execution
* Errors
* Agent handoffs

### Guardrails

Guardrails help constrain the agent and reduce unsafe or unexpected behavior.

They can help control:

* Tool usage
* Input validation
* Output validation
* Maximum iterations
* Permissions
* Security policies

Together:

```text
Agent
  ↓
Orchestration
  ↓
Controlled Execution
  ↓
Guardrails
  ↓
Safer Agent Behavior
```

These topics are explored in more detail in later lessons.

# 13. Practical Example — Flight Booking Agent

Let's make the concepts concrete using the example from the lesson.

The user says:

> "I want to book a flight from Austin to Zurich."

The agent needs to work toward this goal.

The process can be represented as:

```text
User Request
     ↓
"I want to book a flight
from Austin to Zurich."
     ↓
Perceive
     ↓
Reason
     ↓
Find available flights
     ↓
Act
     ↓
Call Flight Search Tool
     ↓
Observe
     ↓
Receive Flight Results
     ↓
Reason Again
     ↓
Enough Information?
     ↓
Yes
     ↓
Final Response
```

---

## Step 1 — Perceive

The agent receives the user's request:

```text
"I want to book a flight from Austin to Zurich."
```

The agent identifies the user's goal.

---

## Step 2 — Reason

The agent determines:

```text
"I need to find available flights."
```

It considers the tools available to it.

It knows that it has access to a **Flight Search Tool**.

Therefore, it decides to use that tool.

---

## Step 3 — Act

The agent calls the flight search tool.

The tool may receive parameters such as:

```text
Origin: Austin
Destination: Zurich
Date: Travel Date
```

Conceptually:

```python
flight_search(
    origin="Austin",
    destination="Zurich",
    date="2026-08-15"
)
```

The exact implementation depends on the agent framework and the available flight service.

---

## Step 4 — Observe

The flight tool returns available results.

For example:

```text
Available Flights:

Flight A — $500
Flight B — $650
Flight C — $800
```

The agent now observes the results.

---

## Step 5 — Reason Again

The agent evaluates the information.

It determines:

```text
"I have enough information to answer the user."
```

It does not need to call another tool.

Therefore, the agent decides to stop the loop.

---

## Step 6 — Final Response

The agent presents the available flight options to the user.

For example:

```text
Here are the available flights from Austin to Zurich:

1. Flight A — $500
2. Flight B — $650
3. Flight C — $800
```

The loop terminates because the agent has enough information to provide a useful response.

# 14. The Most Important Concept

The key idea from this lesson is:

> **The agent decides when to use a tool and when to stop.**

This dynamic decision-making is what makes an AI agent different from a simple LLM call.

A simplified comparison is:

```text
Standalone LLM

Prompt
  ↓
Generate Response
  ↓
Done
```

Versus:

```text
AI Agent

Goal
  ↓
Perceive
  ↓
Reason
  ↓
Should I use a tool?
  ↙             ↘
Yes              No
 ↓                ↓
Use Tool       Final Answer
 ↓
Observe
 ↓
Reason Again
 ↓
Continue or Finish
```

The agent dynamically determines what should happen next.
