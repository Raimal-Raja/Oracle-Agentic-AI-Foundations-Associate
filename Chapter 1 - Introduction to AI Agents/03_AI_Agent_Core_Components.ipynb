# AI Agent Core Components

## Overview

Now that we understand what an AI agent is, this lesson takes a deeper look at the **three core components** that make an LLM-based AI agent work:

1. **LLM — The Brain**
2. **Tools — The Hands**
3. **Loop / Orchestration — The Nervous System**

A useful mental model is:

```text
             AI AGENT
                │
       ┌────────┼────────┐
       ↓        ↓        ↓
      LLM     Tools     Loop
     Brain    Hands   Orchestration
```

These three components work together to create an agentic system.

The lesson emphasizes that even complex agent systems can ultimately be understood by examining these three fundamental components.

---

# The Three Core Components

The fundamental model is:

```text
LLM + Tools + Loop
```

Each component has a different responsibility.

| Component            | Analogy                           | Primary Responsibility              |
| -------------------- | --------------------------------- | ----------------------------------- |
| LLM                  | Brain                             | Reasoning and decision-making       |
| Tools                | Hands                             | Performing external actions         |
| Loop / Orchestration | Nervous System / Operating System | Managing execution and coordination |

A simplified architecture is:

```text
                     User Goal
                         │
                         ▼
                ┌────────────────┐
                │      LLM       │
                │     Brain      │
                │                │
                │ Understand     │
                │ Plan           │
                │ Decide         │
                └───────┬────────┘
                        │
                        ▼
                ┌────────────────┐
                │     Tools      │
                │     Hands      │
                │                │
                │ APIs           │
                │ Databases      │
                │ Code           │
                │ Search         │
                └───────┬────────┘
                        │
                        ▼
                ┌────────────────┐
                │ Loop /          │
                │ Orchestration   │
                │                │
                │ State           │
                │ Routing         │
                │ Memory          │
                │ Safety          │
                └───────┬────────┘
                        │
                        └──────→ Continue / Stop
```

# 1. LLM — The Brain of the Agent

The first core component is the **Large Language Model (LLM)**.

The LLM acts as the **reasoning engine** or **brain** of the agent.

Inside an agent, the LLM performs several important jobs.

It can:

* Understand user intent.
* Plan the steps required to achieve a goal.
* Decide which tool to use.
* Determine the arguments for a tool.
* Interpret tool results.
* Decide what to do next.
* Generate the final response.

Conceptually:

```text
User Request
     ↓
Understand Intent
     ↓
Plan Steps
     ↓
Select Tool
     ↓
Interpret Result
     ↓
Decide Next Step
     ↓
Generate Final Answer
```

The LLM is therefore responsible for the **reasoning and decision-making** that drives the agent.

---

# 2. One or Multiple LLMs

An agent does not necessarily have to use only one LLM.

An agent architecture may use:

* One LLM
* Multiple LLMs
* Different models for different tasks

For example:

```text
                 AI Agent
                    │
        ┌───────────┴───────────┐
        ↓                       ↓
  Fast LLM                 Advanced LLM
        │                       │
Simple Tasks              Complex Reasoning
Routing                    Planning
Classification              Problem Solving
```

Different models may be selected based on:

* Task complexity
* Cost
* Latency
* Accuracy requirements

The same LLM can also power multiple agents with different configurations.

For example:

```text
              Same LLM
                 │
        ┌────────┼────────┐
        ↓        ↓        ↓
     Agent A  Agent B  Agent C
        │        │        │
     Tools A  Tools B  Tools C
```

The agents can have different:

* Instructions
* Prompts
* Tools
* Permissions
* Workflows

This means you do not necessarily need to retrain the LLM for every new agent use case.

---

# 3. Choosing the Right LLM

Choosing an LLM for an agent is a **trade-off problem**.

The goal is not simply to select the model with the highest benchmark score.

Instead, you need to balance:

```text
Capability
    +
Cost
    +
Latency
```

A useful model selection framework is:

```text
             LLM Selection
                  │
       ┌──────────┼──────────┐
       ↓          ↓          ↓
  Capability     Cost      Latency
       │          │          │
       └──────────┼──────────┘
                  ↓
          Right Model for
             the Task
```

Let's examine the major considerations.

---

## 3.1 Instruction Following

An agent's LLM must reliably follow instructions.

The model may receive structured instructions such as:

* System prompts
* Agent role
* Available tools
* Tool usage rules
* Expected output formats
* Safety requirements

For example:

```text
System Instructions:

You are a travel assistant.

Available tools:
- Search Flights
- Search Hotels

Rules:
- Always verify flight availability.
- Never book without user confirmation.
- Return results in a structured format.
```

The LLM must be capable of following these instructions consistently.

---

## 3.2 Context Window

Agents can accumulate a large amount of context during execution.

For example:

```text
User Request
    ↓
Reasoning
    ↓
Tool Call
    ↓
Tool Result
    ↓
Another Reasoning Step
    ↓
Another Tool Call
    ↓
Another Result
```

As the agent executes more steps, additional information may be added to the context.

Therefore, larger context windows can be useful for complex agent workflows.

Conceptually:

```text
Context
├── User Input
├── System Instructions
├── Conversation History
├── Previous Actions
├── Tool Results
├── Observations
└── Current Task State
```

A model with an insufficient context window may struggle with long-running or complex tasks.

---

## 3.3 Reasoning Ability

The reasoning quality of the LLM directly affects the agent's ability to handle complex tasks.

A capable reasoning model may perform better at:

* Planning
* Multi-step problem solving
* Tool selection
* Decision-making
* Error recovery
* Complex task decomposition

For example:

```text
Simple Task
    ↓
Basic Reasoning
    ↓
Fast Model May Be Enough
```

While:

```text
Complex Multi-Step Task
    ↓
Advanced Reasoning
    ↓
More Capable Model May Be Required
```

---

## 3.4 Cost

Every iteration of an agent may involve one or more LLM API calls.

For example:

```text
Iteration 1 → LLM Call
Iteration 2 → LLM Call
Iteration 3 → LLM Call
Iteration 4 → LLM Call
```

More iterations can lead to:

* Higher token usage
* Higher API costs
* Greater infrastructure costs

Therefore, cost is an important consideration when designing agents.

---

## 3.5 Latency

Each LLM call takes time.

An agent that performs many iterations may have higher latency.

For example:

```text
LLM Call
    ↓
Tool Call
    ↓
LLM Call
    ↓
Tool Call
    ↓
LLM Call
    ↓
Final Response
```

This can take longer than a simple:

```text
Prompt
    ↓
LLM
    ↓
Response
```

Therefore, developers need to balance model capability with response speed.

---

# 4. Two-Tier Model Strategy

Some systems use different models for different levels of complexity.

For example:

```text
                 User Request
                      ↓
                 Routing Model
                      │
             ┌────────┴────────┐
             ↓                 ↓
        Simple Task       Complex Task
             ↓                 ↓
       Fast / Cheap       Powerful Model
          Model              Model
             ↓                 ↓
          Result            Result
```

A cheaper and faster model can handle:

* Simple routing
* Classification
* Basic decisions

A more capable model can handle:

* Complex reasoning
* Multi-step planning
* Difficult problem solving

This approach can reduce costs while maintaining high-quality results where advanced reasoning is necessary.

---

# 5. Reasoning Models

Some modern reasoning models are designed or trained to perform more sophisticated internal reasoning.

These models may improve performance on complex agent tasks.

However, they may also introduce:

* Higher cost
* Higher latency

Therefore, the decision should still be based on the application's requirements.

The key lesson is:

> **The best LLM for an agent is not always the most powerful LLM available.**

Instead, model selection should balance:

```text
Capability
Cost
Latency
Context
Instruction Following
Reasoning Quality
```

---

# 6. Tools — The Hands of the Agent

The second core component is **Tools**.

Tools give an AI agent the ability to interact with the real world.

A useful analogy is:

> **The LLM is the brain, and tools are the hands.**

Without tools:

```text
LLM
 ↓
Generate Text
```

With tools:

```text
LLM
 ↓
Decide What to Do
 ↓
Use Tool
 ↓
Perform Real-World Action
 ↓
Observe Result
```

Tools can allow an agent to:

* Query databases.
* Call APIs.
* Execute code.
* Search the web.
* Access files.
* Retrieve information.
* Perform business operations.
* Interact with enterprise systems.

Tools transform an LLM from primarily a **text generator** into an **actor capable of performing actions**.

---

# 7. How Function Calling and Tool Calling Work

Tool calling is a fundamental concept in agentic AI.

The general process is:

```text
1. Define Available Tools
        ↓
2. Send Tool Definitions to LLM
        ↓
3. LLM Analyzes User Request
        ↓
4. LLM Decides Whether Tool Is Needed
        ↓
5. LLM Generates Structured Tool Call
        ↓
6. Agent Validates Tool Call
        ↓
7. Agent Executes Tool
        ↓
8. Tool Returns Result
        ↓
9. Agent Sends Result to LLM
        ↓
10. LLM Decides What to Do Next
```

This creates a communication loop between the LLM and the tools.

---

# 8. Tool Definitions and JSON Schemas

Tools are often described to an LLM using structured schemas.

Conceptually, a tool definition might look like:

```json
{
  "name": "get_weather",
  "description": "Get the current weather for a city",
  "parameters": {
    "type": "object",
    "properties": {
      "city": {
        "type": "string",
        "description": "Name of the city"
      }
    },
    "required": ["city"]
  }
}
```

The LLM receives information about:

* Tool name
* Tool purpose
* Required parameters
* Parameter types

The LLM can then decide whether the tool is appropriate for the user's request.

---

# 9. Example — Weather Tool Calling

Suppose the user asks:

> "What is the current temperature in Paris?"

The agent has access to a weather tool.

The process may look like this:

```text
User:
"What is the current temperature in Paris?"
        ↓
LLM receives:
- User request
- Available tool definitions
        ↓
LLM decides:
"I need current weather information."
        ↓
LLM creates structured tool call:
get_weather(city="Paris")
        ↓
Agent receives tool call
        ↓
Agent validates request
        ↓
Agent executes weather API
        ↓
Weather API returns result
        ↓
Agent sends result to LLM
        ↓
LLM generates final answer
```

The final answer might be:

```text
The current temperature in Paris is 22°C.
```

---

# 10. Critical Security Boundary

One of the most important concepts in tool calling is:

> **The LLM does not execute the tool itself.**

The LLM only generates a **structured tool call request**.

The actual execution is performed by the agent application or the underlying code.

The architecture is:

```text
             LLM
              │
              │ Structured Tool Call
              ▼
       Agent Application
              │
              │ Validate
              ▼
         Tool Executor
              │
              ▼
        External System
```

For example:

```text
LLM
 ↓
"Call get_weather with city=Paris"
 ↓
Agent Code
 ↓
Validate Request
 ↓
Execute Weather API
 ↓
Return Result
```

This separation creates an important **security boundary**.

The LLM can request an action, but the application determines whether that action is:

* Valid
* Authorized
* Safe
* Allowed

This is a critical principle in production agent systems.

---

# 11. LLM vs. Agent Code Responsibilities

It is useful to clearly separate the responsibilities of the LLM and the agent application.

| Component  | Responsibility          |
| ---------- | ----------------------- |
| LLM        | Reasoning               |
| LLM        | Tool selection          |
| LLM        | Tool arguments          |
| LLM        | Interpreting results    |
| Agent Code | Validating tool calls   |
| Agent Code | Executing tools         |
| Agent Code | Enforcing permissions   |
| Agent Code | Applying security rules |

The key idea is:

```text
LLM = Decides What to Request
Agent Code = Decides Whether and How to Execute
```

This distinction is essential for secure agent design.

---

# 12. Model Context Protocol (MCP)

The lesson also introduces the **Model Context Protocol (MCP)**.

MCP provides a standardized approach for defining and connecting tools and other external capabilities to AI applications.

Conceptually:

```text
Without Standardization

LLM A → Custom Tool Interface A
LLM B → Custom Tool Interface B
LLM C → Custom Tool Interface C
```

With standardized interfaces:

```text
                  MCP
                   │
       ┌───────────┼───────────┐
       ↓           ↓           ↓
    Tool A      Tool B      Tool C
       │           │           │
       └───────────┼───────────┘
                   ↓
             AI Applications
```

The course will explore MCP in greater detail in a dedicated module.

The important idea at this stage is that MCP aims to make it easier to connect AI applications with external tools and context using standardized interfaces.

# 13. Loop / Orchestration — The Nervous System

The third core component is the **Loop**, also called **Orchestration**.

A useful analogy is:

> **The loop is the agent's nervous system.**

Another useful analogy is:

> **The orchestration layer is like an operating system for the agent.**

The loop brings everything together.

It manages:

* Reasoning cycles
* State
* Memory
* Tool routing
* Flow control
* Safety rules
* Errors
* Retries
* Termination

A simplified architecture is:

```text
                    Agent
                      │
              ┌───────┴───────┐
              ↓               ↓
             LLM            Tools
              │               │
              └───────┬───────┘
                      ↓
                Orchestration
                      │
          ┌───────────┼───────────┐
          ↓           ↓           ↓
        State       Memory      Safety
          │           │           │
          └───────────┼───────────┘
                      ↓
                Continue / Stop
```

---

# 14. Orchestration as an Application Controller

In a traditional application, an application controller may handle:

* Routing
* State management
* Flow control
* Error handling

The orchestration layer in an AI agent performs a similar role.

However, instead of routing only HTTP requests, it manages:

* Reasoning steps
* Tool calls
* Observations
* Agent state
* Execution flow

The comparison is:

```text
Traditional Application
        ↓
Application Controller
        ↓
Routes Requests
        ↓
Manages State
        ↓
Controls Flow
```

Compared with:

```text
AI Agent
        ↓
Orchestration Layer
        ↓
Routes Reasoning
        ↓
Routes Tool Calls
        ↓
Manages State
        ↓
Controls Agent Loop
```

# 15. Core Responsibilities of the Orchestration Layer

The lesson introduces several important responsibilities of the orchestration layer.

These include:

1. Loop Management
2. Reasoning Strategy
3. Memory Management
4. Tool Routing
5. State Machine
6. Safety and Guardrails

Let's examine each.

---

## 15.1 Loop Management

The orchestration layer manages the agent execution cycle.

The cycle may look like:

```text
Reason
  ↓
Act
  ↓
Observe
  ↓
Reason Again
  ↓
Continue / Pause / Stop
```

The orchestration layer decides:

* When to continue.
* When to pause.
* When to stop.
* When to retry.

It can also enforce limits such as:

* Maximum iterations
* Maximum execution time
* Maximum tool calls

---

## 15.2 Reasoning Strategy

The orchestration layer can apply structured reasoning strategies.

Examples include:

* ReAct
* Chain of Thought

These approaches can help break complex goals into smaller tasks.

Conceptually:

```text
Complex Goal
     ↓
Reasoning Strategy
     ↓
Task Decomposition
     ↓
Actionable Steps
     ↓
Tool Execution
```

The exact reasoning implementation depends on the model and agent framework.

---

## 15.3 Memory Management

Agents may need to maintain context across multiple interactions.

Memory can include:

### Short-Term Working Memory

Information needed during the current task.

For example:

```text
Current User Request
Current Tool Results
Current Task State
Recent Conversation
```

### Long-Term Knowledge

Information that needs to persist beyond a single interaction.

For example:

```text
User Preferences
Historical Information
Enterprise Knowledge
Stored Documents
Database Records
```

The orchestration layer can manage how this information is stored and retrieved.

Conceptually:

```text
Agent
  │
  ├── Short-Term Memory
  │
  └── Long-Term Knowledge
```

---

## 15.4 Tool Routing

The orchestration layer helps determine which tool should be used for a particular step.

For example:

```text
User Request
     ↓
Need Current Weather
     ↓
Weather Tool
```

Or:

```text
User Request
     ↓
Need Customer Information
     ↓
Customer Database Tool
```

The orchestration layer is responsible for coordinating tool usage, including:

* Selecting tools
* Formatting requests
* Validating requests
* Handling failures
* Processing results

---

## 15.5 State Machine

An agent can be modeled as a state machine.

The state machine tracks where the agent is in the task.

For example:

```text
START
  ↓
UNDERSTAND_REQUEST
  ↓
PLAN
  ↓
CALL_TOOL
  ↓
OBSERVE_RESULT
  ↓
TASK_COMPLETE?
  ├── No → PLAN
  │
  └── Yes
        ↓
      FINISH
```

The agent may also enter states such as:

```text
ERROR
RETRY
WAITING
ESCALATE
CANCELLED
```

Tracking state helps the system determine what should happen next.

---

## 15.6 Safety and Guardrails

The orchestration layer can enforce system-level rules that override model decisions.

For example:

```text
LLM requests action
        ↓
Safety Check
        ↓
Is action allowed?
     ↙       ↘
   Yes        No
    ↓          ↓
Execute     Block
```

Guardrails can help prevent:

* Unsafe actions
* Invalid actions
* Unauthorized actions
* Excessive tool usage
* Infinite loops
* Policy violations

This is particularly important when an agent has access to powerful tools.

---

The three core components work together.

```text
                  AI AGENT
                     │
     ┌───────────────┼───────────────┐
     ↓               ↓               ↓
    LLM            Tools           Loop
   Brain           Hands       Orchestration
     │               │               │
     │               │               │
Reasoning        Actions         Coordination
Planning         APIs            State
Decisions        Database        Memory
Intent           Code            Safety
     │               │               │
     └───────────────┼───────────────┘
                     ↓
              Agentic Behavior
```

The LLM decides what should happen.

Tools perform external actions.

The orchestration layer coordinates the entire process.

A useful analogy from the lesson is to think of the orchestration layer as the **conductor of an agentic symphony**.

In an orchestra:

* Different musicians have different roles.
* The conductor coordinates them.
* The conductor controls timing.
* The conductor determines when each section participates.

Similarly, in an AI agent:

* The LLM provides reasoning.
* Tools perform actions.
* Memory provides context.
* State tracks progress.
* Guardrails enforce rules.
* Orchestration coordinates everything.

The orchestration layer decides:

> When should the agent reason?

> Which tool should be called?

> What should happen with the result?

> Should the agent continue?

> Should it stop?

> Should it retry?

> Should it escalate to a human?

The complete conceptual architecture can be represented as:

This architecture demonstrates how the three components cooperate.

  * Capability
  * Cost

  * Loop execution
  * Reasoning strategies

  * Tool routing

# Quick Revision Questions

### 1. What are the three core components of an AI agent?

The three core components are:

```text
LLM + Tools + Loop / Orchestration
```

### 2. What is the role of the LLM?

The LLM acts as the reasoning engine. It understands user intent, plans tasks, selects tools, interprets results, and generates final responses.

### 3. What factors should be considered when selecting an LLM?

Important factors include:

* Capability
* Cost
* Latency
* Instruction following
* Context window
* Reasoning ability

### 4. Why are tools important?

Tools enable agents to interact with external systems and perform actions such as querying databases, calling APIs, executing code, and searching for information.

### 5. Does the LLM execute tools directly?

**No.**

The LLM generates a structured tool-call request. The agent application or tool execution layer validates and executes the actual function.

### 6. Why is tool execution a security boundary?

Because the LLM only requests an action. The application can validate the request, check permissions, enforce policies, and decide whether the action is allowed before execution.

### 7. What is the role of the orchestration layer?

The orchestration layer coordinates the agent's execution, manages state and memory, routes tools, controls the loop, handles errors, and applies safety rules.

### 8. What is tool calling?

Tool calling is the process where an LLM receives information about available tools, selects an appropriate tool, generates structured arguments, and requests that the agent execute the tool.

### 9. What is MCP?

**Model Context Protocol (MCP)** is a standardized approach for connecting AI applications and agents with tools and external context.

### 10. What is the agent loop responsible for?

The agent loop manages the repeated execution cycle of:

```text
Reason → Act → Observe → Reason Again
```

It determines whether to continue, stop, retry, or escalate.

### 11. What is a state machine in an agent?

A state machine tracks the agent's progress through different states of a task, such as planning, tool execution, observation, completion, retry, or failure.

### 12. Why are guardrails important?

Guardrails enforce system-level rules and constraints that help prevent unsafe, invalid, or unauthorized actions.

# Final Summary

An LLM-based AI agent can be understood through three fundamental components:

```text
             ┌─────────────┐
             │     LLM     │
             │    Brain    │
             └──────┬──────┘
                    │
                    ▼
             ┌─────────────┐
             │    Tools    │
             │    Hands    │
             └──────┬──────┘
                    │
                    ▼
             ┌─────────────┐
             │    Loop     │
             │Orchestration│
             │   Nervous   │
             │   System    │
             └─────────────┘
```

The **LLM** provides the intelligence and reasoning.

The **tools** provide the ability to interact with the real world.

The **loop or orchestration layer** coordinates everything and manages the execution process.

The overall model is:

> **LLM = Thinks**

> **Tools = Acts**

> **Orchestration = Coordinates**

Together, these components transform a language model into an agentic system capable of reasoning about goals, selecting tools, executing actions, observing results, managing state, and continuing until the task is complete.

The most important conceptual model to remember is:

```text
LLM
  ↓
Reason
  ↓
Choose Tool
  ↓
Tool Executes
  ↓
Observe Result
  ↓
Orchestration Manages State
  ↓
Reason Again
  ↓
Continue / Retry / Stop
```

This three-component architecture forms the foundation for understanding more advanced agentic AI frameworks and technologies introduced later in the course.
