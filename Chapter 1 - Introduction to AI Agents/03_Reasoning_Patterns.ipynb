# Reasoning Patterns for AI Agents

## Introduction

AI agents need to do more than simply generate text. They need to **reason, plan, decide what to do next, use tools, and adapt based on results**.

In this lesson, we explore three important reasoning patterns:

1. **Chain of Thought (CoT)**
2. **ReAct (Reasoning + Acting)**
3. **Tree of Thought (ToT)**

These approaches provide different ways for AI systems to solve complex problems.

The most commonly used pattern in practical AI agent systems is **ReAct**, because it combines reasoning with external tool usage in an iterative loop.

# 1. What Are Reasoning Patterns?

A reasoning pattern defines how an AI model or agent approaches a problem.

A basic LLM interaction might look like:

```text
User Question
      ↓
      LLM
      ↓
Final Answer
```

This works well for simple questions.

However, complex tasks may require multiple steps:

```text
User Goal
    ↓
Understand the Goal
    ↓
Reason About the Problem
    ↓
Plan Next Action
    ↓
Use a Tool
    ↓
Observe the Result
    ↓
Reason Again
    ↓
Final Answer
```

Reasoning patterns help an agent determine:

* How to approach a problem.
* How to break down a complex task.
* When to use a tool.
* What action to take next.
* When enough information has been collected.
* When to provide the final answer.

The three major reasoning patterns discussed in this lesson are:

```text
                    Reasoning Patterns
                           │
          ┌────────────────┼────────────────┐
          ↓                ↓                ↓
   Chain of Thought      ReAct        Tree of Thought
          │                │                │
   Step-by-Step      Reason + Act     Multiple Paths
     Reasoning       + Observe        Exploration
```

# 2. Chain of Thought

## What Is Chain of Thought?

**Chain of Thought**, often abbreviated as **CoT**, is a prompting technique that encourages an LLM to reason through a problem step by step.

The core idea is:

> Instead of asking only for the final answer, encourage the model to work through the problem step by step.

A simple question might be:

```text
What is the answer?
```

A Chain of Thought-style instruction might be:

```text
Think step by step and solve the problem.
```

The conceptual flow becomes:

```text
Problem
   ↓
Step 1
   ↓
Step 2
   ↓
Step 3
   ↓
Final Answer
```

Chain of Thought is especially useful for:

* Mathematics
* Logic
* Puzzles
* Structured analysis
* Multi-step reasoning

---

# 3. Chain of Thought Example

Consider this problem:

> A store has 15 apples and sells 3 batches of 4 apples. How many apples are left?

Without explicitly requesting step-by-step reasoning, the model may simply answer:

```text
3 apples
```

With a Chain of Thought-style approach, the reasoning can be represented as:

```text
The store has 15 apples.

There are 3 batches.

Each batch contains 4 apples.

Total apples sold:
3 × 4 = 12

Remaining apples:
15 - 12 = 3

Answer:
3 apples remain.
```

The final answer is the same, but the reasoning process is more explicit and traceable.

The important point is:

> **Chain of Thought does not necessarily make the model smarter. It helps the model use its reasoning capabilities more reliably.**

---

# 4. Zero-Shot Chain of Thought

One variation of Chain of Thought is called **Zero-Shot Chain of Thought**.

The model is not given examples of reasoning. Instead, the prompt simply encourages step-by-step thinking.

For example:

```text
Question:
If a store has 15 apples and sells 3 batches of 4 apples,
how many apples are left?

Let's think step by step.
```

The model is encouraged to reason through the problem before producing the answer.

Conceptually:

```text
Question
   ↓
"Let's think step by step"
   ↓
Intermediate Reasoning
   ↓
Final Answer
```

The key idea is that no reasoning examples are provided beforehand.

---

# 5. Few-Shot Chain of Thought

Another variation is **Few-Shot Chain of Thought**.

In this approach, the prompt contains examples that demonstrate how a problem should be solved.

For example:

```text
Example:

Question:
John has 10 apples and gives away 3.

Reasoning:
John starts with 10 apples.
He gives away 3 apples.
10 - 3 = 7.

Answer:
7 apples remain.
```

Then the model receives a new problem.

The model can use the examples as a pattern for solving the new problem.

The conceptual flow is:

```text
Example 1
     ↓
Example 2
     ↓
Example 3
     ↓
New Problem
     ↓
Follow the Demonstrated Reasoning Pattern
     ↓
Answer
```

---

# 6. Benefits of Chain of Thought

Chain of Thought provides several potential benefits.

## 6.1 Improved Reasoning

Breaking a complex problem into smaller steps can help the model reason more effectively.

Instead of:

```text
Question → Answer
```

The model follows:

```text
Question
   ↓
Intermediate Steps
   ↓
Reasoning
   ↓
Answer
```

---

## 6.2 Transparency

Chain of Thought can make the reasoning process more visible and understandable.

For example:

```text
Problem
   ↓
Step 1
   ↓
Step 2
   ↓
Step 3
   ↓
Answer
```

This makes it easier to understand how the solution was reached.

---

## 6.3 Debuggability

If the final answer is incorrect, the reasoning process may help identify where the mistake occurred.

For example:

```text
Step 1 → Correct
Step 2 → Correct
Step 3 → Incorrect
Step 4 → Incorrect Result
```

This can help developers identify the source of an error.

---

# 7. Limitations of Chain of Thought

Standalone Chain of Thought has an important limitation.

The model primarily relies on:

* Its internal knowledge.
* Its learned patterns.
* Its reasoning capabilities.

It cannot automatically access external information.

For example, suppose the user asks:

```text
What is the current weather in London?
```

Chain of Thought alone cannot retrieve live weather data.

Similarly:

```text
What is today's stock price for a company?
```

The model cannot reliably provide current information without access to an external data source.

The limitations include:

* No direct access to external tools.
* No automatic web search.
* No live information retrieval.
* No external verification.
* Limited ability to self-correct using real-world data.

Conceptually:

```text
User Question
      ↓
LLM
      ↓
Internal Knowledge
      ↓
Reasoning
      ↓
Answer
```

The model has no mechanism to check whether its internal knowledge is accurate or current.

This creates a risk:

```text
Incorrect Internal Knowledge
          ↓
Confident Reasoning
          ↓
Incorrect Answer
```

This limitation is one reason why tool-using AI agents are important.

# 8. ReAct

## What Is ReAct?

**ReAct** stands for:

> **Reasoning + Acting**

ReAct combines reasoning with external tool use.

The basic idea is:

```text
Reason
   ↓
Act
   ↓
Observe
   ↓
Reason Again
   ↓
Act Again
   ↓
Observe Again
   ↓
Final Answer
```

A simple way to remember ReAct is:

> **ReAct = Chain of Thought + Tool Use + Loop**

This makes ReAct particularly useful for AI agents.

---

# 9. ReAct Example — Flight Search

Suppose a user says:

> "I want to find flights from Austin to Zurich."

The agent needs to determine how to complete this request.

## Step 1: Reason

The agent reasons:

```text
I need to find available flights
from Austin to Zurich.
```

The agent recognizes that it needs external information.

---

## Step 2: Select a Tool

The agent decides:

```text
I should use the flight search tool.
```

---

## Step 3: Act

The agent generates a structured tool call:

```python
search_flights(
    origin="Austin",
    destination="Zurich",
    date="user_provided_date"
)
```

The tool is then executed by the agent application.

---

## Step 4: Observe

The tool returns results:

```text
Flight 1 → $500
Flight 2 → $650
Flight 3 → $800
```

The agent observes the results.

---

## Step 5: Reason Again

The agent now has enough information.

It reasons:

```text
I have the available flight results.

I should present the options to the user.
```

---

## Step 6: Final Answer

The agent provides the results:

```text
Available flights from Austin to Zurich:

Flight 1: $500
Flight 2: $650
Flight 3: $800
```

The complete ReAct flow is:

```text
User Request
      ↓
REASON
"I need to find available flights."
      ↓
ACT
Call Flight Search Tool
      ↓
OBSERVE
Receive Flight Results
      ↓
REASON
Analyze the Results
      ↓
FINAL ANSWER
Present Flight Options
```

This is the core ReAct pattern.

---

# 10. Why ReAct Is Important

ReAct solves a fundamental limitation of Chain of Thought.

Chain of Thought provides:

```text
Reasoning
```

But it does not automatically provide:

```text
External Tool Access
```

On the other hand, tool use provides:

```text
Action
```

But the tool itself does not decide:

```text
When should I use this tool?
Why should I use this tool?
What should I do after getting the result?
```

ReAct combines both.

```text
Reasoning
     +
Tool Use
     +
Observation
     +
Iteration
```

This allows an agent to reason about:

* Whether a tool is necessary.
* Which tool should be used.
* What arguments should be passed.
* What to do with the tool result.
* Whether another tool call is needed.
* When the task is complete.

---

# 11. ReAct as an Agent Execution Loop

A ReAct agent can be visualized as:

```text
                User Request
                     ↓
                  Reason
                     ↓
              Need a Tool?
                ↙       ↘
              Yes        No
               ↓          ↓
          Call Tool    Final Answer
               ↓
            Observe
               ↓
        Reason About Result
               ↓
        Need More Information?
             ↙       ↘
           Yes        No
            ↓          ↓
       Call Tool    Final Answer
            │
            └──────→ Observe
```

This loop continues until:

* The agent has enough information.
* The task is completed.
* A safety limit is reached.
* The agent needs human assistance.

# 12. ReAct and Transparency

One advantage of ReAct is that the agent's actions can be made observable.

For example:

```text
User Request:
Find the cheapest flight.

Agent:
I need current flight information.

Tool Selected:
Flight Search API

Tool Input:
Origin = Austin
Destination = Zurich

Tool Result:
Flight A = $500
Flight B = $650
Flight C = $800

Agent:
The cheapest flight is Flight A.

Final Answer:
Flight A costs $500.
```

This type of execution trace can help with:

* Auditing.
* Debugging.
* Monitoring.
* Error analysis.
* Understanding tool usage.

However, it is important to distinguish between **internal model reasoning** and **observable execution traces**.

Modern AI systems may keep detailed internal reasoning private while exposing only:

* Tool calls.
* Tool results.
* Short summaries.
* Final responses.

# 14. ReAct Prevents Unnecessary Guessing

One of the biggest advantages of ReAct is that the agent can decide when it needs external information.

For example:

```text
User:
What is the current temperature in Paris?
```

The agent reasons:

```text
I need current information.

I should use the weather tool.
```

Then:

```text
Weather Tool
      ↓
22°C
      ↓
Agent
      ↓
The current temperature in Paris is 22°C.
```

Instead of relying on potentially outdated internal knowledge, the agent retrieves current information.

The conceptual difference is:

### Chain of Thought Only

```text
Question
   ↓
Internal Knowledge
   ↓
Reasoning
   ↓
Answer
```

### ReAct

```text
Question
   ↓
Reason
   ↓
Use Tool
   ↓
Get Current Data
   ↓
Observe
   ↓
Reason
   ↓
Answer
```

# 15. Interleaving Reasoning and Acting

A key concept in ReAct is **interleaving**.

Interleaving means that reasoning and actions are performed alternately.

The agent does not simply:

```text
Reason Once
   ↓
Perform All Actions
   ↓
Answer
```

Instead, it does:

```text
Reason
   ↓
Act
   ↓
Observe
   ↓
Reason
   ↓
Act
   ↓
Observe
   ↓
Reason
   ↓
Answer
```

This allows the agent to adapt to new information.

For example:

```text
User:
Find me the cheapest flight to Zurich.
```

The agent may:

```text
Reason:
I need flight prices.
        ↓
Act:
Search for flights.
        ↓
Observe:
Three flights found.
        ↓
Reason:
I need to compare their prices.
        ↓
Act:
Analyze the results.
        ↓
Observe:
Cheapest flight costs $500.
        ↓
Answer:
The cheapest flight is $500.
```

The agent dynamically decides what to do next based on observations.

# 16. Tree of Thought

The third reasoning pattern is **Tree of Thought**, abbreviated as **ToT**.

Tree of Thought is more complex than a linear Chain of Thought approach.

Instead of following one reasoning path, the model explores multiple possible approaches.

Conceptually:

```text
                  Problem
                     │
          ┌──────────┼──────────┐
          ↓          ↓          ↓
      Approach A  Approach B  Approach C
          │          │          │
       Explore    Explore    Explore
          │          │          │
          └──────────┼──────────┘
                     ↓
              Select Best Path
                     ↓
                  Answer
```

The agent effectively considers:

```text
What happens if I choose option A?

What happens if I choose option B?

What happens if I choose option C?
```

Then it evaluates the alternatives.

---

# 17. Tree of Thought Analogy — Chess

A useful analogy is a chess player.

A chess player may think:

```text
If I move here...
    ↓
The opponent might respond this way...
    ↓
Then I can respond with this move.
```

At the same time, the player might consider another possibility:

```text
If I move there...
    ↓
The opponent may respond differently...
    ↓
I could then make another move.
```

The player explores multiple possibilities before choosing a move.

This is similar to Tree of Thought.

```text
                  Current Position
                         │
              ┌──────────┼──────────┐
              ↓          ↓          ↓
           Move A      Move B      Move C
              │          │          │
           Evaluate    Evaluate    Evaluate
              │          │          │
              └──────────┼──────────┘
                         ↓
                   Best Decision
```

---

# 18. Chain of Thought vs. Tree of Thought

## Chain of Thought

Follows a mostly linear reasoning path.

```text
Problem
   ↓
Thought 1
   ↓
Thought 2
   ↓
Thought 3
   ↓
Answer
```

## Tree of Thought

Explores multiple reasoning paths.

```text
                  Problem
                     │
           ┌─────────┼─────────┐
           ↓         ↓         ↓
        Path A     Path B     Path C
           │         │         │
        Explore   Explore   Explore
           │         │         │
           └─────────┼─────────┘
                     ↓
               Best Solution
```

The main difference is:

> **Chain of Thought follows a reasoning chain, while Tree of Thought explores a reasoning tree.**

# 19. Comparison of Reasoning Patterns

| Feature                  | Chain of Thought   | ReAct         | Tree of Thought      |
| ------------------------ | ------------------ | ------------- | -------------------- |
| Step-by-step reasoning   | Yes                | Yes           | Yes                  |
| External tool use        | No                 | Yes           | Optional             |
| Iterative loop           | No                 | Yes           | Can be               |
| Multiple reasoning paths | No                 | Usually not   | Yes                  |
| External information     | No                 | Yes           | Optional             |
| Best for                 | Math and logic     | Agentic tasks | Complex exploration  |
| Main concept             | Think step by step | Reason + Act  | Explore alternatives |

A simplified comparison:

```text
Chain of Thought
Question
    ↓
Reason
    ↓
Answer
```

```text
ReAct
Question
    ↓
Reason
    ↓
Act
    ↓
Observe
    ↓
Reason
    ↓
Answer
```

```text
Tree of Thought
Question
    ↓
Explore Path A
Explore Path B
Explore Path C
    ↓
Evaluate
    ↓
Select Best Path
    ↓
Answer
```

# 20. Why ReAct Is Common in Production Agents

In practice, many production agents use patterns similar to ReAct.

The reason is that ReAct provides a useful balance between:

* Reasoning capability.
* External tool use.
* Adaptability.
* Transparency.
* Cost.

A production agent can follow a process such as:

```text
User Request
      ↓
LLM Reasoning
      ↓
Does it need a tool?
    ↙       ↘
  Yes        No
   ↓          ↓
Tool Call   Final Answer
   ↓
Tool Result
   ↓
LLM Reasoning
   ↓
Need another tool?
    ↙       ↘
  Yes        No
   ↓          ↓
Tool Call   Final Answer
```

The agent continues until the task is complete.

# 21. ReAct in Agent Frameworks

Modern agent frameworks commonly implement concepts similar to the ReAct pattern.

Examples include:

* **LangChain**

A simplified framework flow is:

```text
System Prompt
      ↓
Available Tools
      ↓
User Request
      ↓
LLM Reasoning
      ↓
Tool Call
      ↓
Tool Execution
      ↓
Tool Result
      ↓
LLM Reasoning
      ↓
Final Response
```

The system instructions may conceptually tell the model:

```text
You have access to these tools.

Think about the task.

If you need external information,
use an appropriate tool.

When you have enough information,
provide the final answer.
```

The framework manages the execution process.

# 22. Practical Conceptual Python Example

The following example demonstrates a simplified ReAct-style workflow.

> **Note:** This is a conceptual implementation. It does not implement a complete production agent or expose private model reasoning. It demonstrates the basic architecture of reasoning, tool use, observation, and final response.

```python
def search_flights(origin, destination):
    """
    Simulated external tool.
    In a real application, this could call
    a flight booking API.
    """
    return [
        {"flight": "Flight A", "price": 500},
        {"flight": "Flight B", "price": 650},
        {"flight": "Flight C", "price": 800},
    ]


def react_agent(user_request):
    # Reason
    print("Agent needs flight information.")

    # Act
    flights = search_flights(
        origin="Austin",
        destination="Zurich"
    )

    # Observe
    print("Tool result:", flights)

    # Reason about the result
    cheapest = min(
        flights,
        key=lambda flight: flight["price"]
    )

    # Final answer
    return (
        f"The cheapest flight is "
        f"{cheapest['flight']} "
        f"at ${cheapest['price']}."
    )


answer = react_agent(
    "Find the cheapest flight from Austin to Zurich."
)

print(answer)
```

The architecture is:

```text
User Request
      ↓
Reason
      ↓
search_flights()
      ↓
Observe Results
      ↓
Analyze Results
      ↓
Final Answer
```

In a real AI agent, an LLM would dynamically determine:

* Whether a tool is required.
* Which tool should be called.
* What arguments should be provided.
* Whether another tool should be called.
* When the task is complete.

# 23. A More General ReAct Pseudocode

A generic ReAct agent can be represented as:

```python
while not task_complete:

    # Reason
    decision = llm(user_request, available_tools)

    if decision.requires_tool:

        # Act
        result = execute_tool(
            decision.tool,
            decision.arguments
        )

        # Observe
        context.append(result)

    else:

        # Final answer
        return decision.final_answer
```

The basic pattern is:

```text
while task is not complete:

    Reason

    if tool is required:
        Act
        Observe
    else:
        Return Final Answer
```

This is a simplified representation of the agent execution loop.
