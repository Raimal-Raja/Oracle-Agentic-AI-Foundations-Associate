# Building Your First AI Agent

## Introduction

In this lesson, we build our first AI agent and learn how agent frameworks simplify the development of agentic AI applications.

The key idea is that building a basic AI agent can be surprisingly simple. A minimal agent can be created by combining:

1. An **LLM (Large Language Model)** as the agent's brain.
2. **Tools** that allow the agent to perform actions.
3. An **Agent Framework** that manages the reasoning and execution loop.
4. A way to **invoke the agent** with a user request.

The agent framework handles the underlying **ReAct-style loop**:

```text
Reason → Act → Observe → Reason Again → Final Answer
```

The agent dynamically decides which tools to use and how to combine their results.

# 1. What Is an AI Agent Framework?

Building an AI agent from scratch requires implementing several components:

* LLM interaction.
* Tool definitions.
* Tool selection.
* Tool execution.
* State management.
* Reasoning loops.
* Error handling.
* Guardrails.
* Memory.
* Observability.

Instead of implementing everything manually, developers can use **agent frameworks**.

An agent framework provides abstractions and utilities for building AI agents.

A simplified architecture looks like:

```text
                 User
                  │
                  ↓
           Agent Framework
                  │
        ┌─────────┴─────────┐
        ↓                   ↓
       LLM                Tools
        │                   │
        └─────────┬─────────┘
                  ↓
            Agent Loop
                  │
                  ↓
           Final Response
```

The framework manages the interaction between the LLM and the available tools.

# 3. LangChain

**LangChain** is one of the most widely adopted frameworks for developing applications powered by LLMs and agents.

The name originally comes from the concept of **chaining LLM calls together**.

For example:

```text
LLM Call 1
    ↓
Result
    ↓
LLM Call 2
    ↓
Result
    ↓
LLM Call 3
```

LangChain provides a large ecosystem of:

* Integrations.
* Tools.
* Models.
* APIs.
* Agent abstractions.
* Memory components.
* Middleware.
* Testing utilities.

## Strengths

LangChain is useful because it provides:

* A large ecosystem.
* Many integrations.
* Extensive documentation.
* Support for complex agent applications.
* Production-oriented features.

## Weaknesses

For very simple applications, LangChain can sometimes feel over-engineered.

Its abstractions can introduce additional complexity.

It is particularly useful for production teams that need:

* Orchestration.
* State management.
* Middleware.
* Testing.
* Integrations.
* Complex workflows.

# 4. LangGraph

**LangGraph** extends the agent development ecosystem with stateful, graph-based workflows.

Instead of thinking only in terms of a linear chain:

```text
Step 1
  ↓
Step 2
  ↓
Step 3
```

A graph-based workflow can contain multiple paths:

```text
              Start
                ↓
          ┌─────┴─────┐
          ↓           ↓
       Step A       Step B
          ↓           ↓
          └─────┬─────┘
                ↓
             Step C
```

This makes graph-based approaches useful for:

* Stateful agents.
* Complex workflows.
* Multi-step processes.
* Conditional routing.
* Human-in-the-loop systems.
* Customized orchestration.

LangGraph provides more control when the workflow structure becomes complex.

# 12. Python Tool Example

A simplified mathematical tool might look like:

```python
def add(a: float, b: float) -> float:
    """
    Add two numbers together.
    """
    return a + b
```

Another tool:

```python
def multiply(a: float, b: float) -> float:
    """
    Multiply two numbers together.
    """
    return a * b
```

Division:

```python
def divide(a: float, b: float) -> float:
    """
    Divide the first number by the second number.
    """
    return a / b
```

These functions are simple Python functions.

---

# 13. Tool Decorators

Agent frameworks often provide a **tool decorator**.

Conceptually:

```python
@tool
def add(a: float, b: float) -> float:
    """
    Add two numbers together.
    """
    return a + b
```

The decorator tells the framework:

> Register this Python function as a tool that an AI agent can discover and call.

Without the decorator, the framework may treat the function as an ordinary Python function.

With the decorator:

```text
Python Function
      ↓
Tool Decorator
      ↓
Registered Tool
      ↓
Available to Agent
```

The agent can now potentially select the tool when it determines that the tool is useful.

---

# 14. Type Hints

Type hints are important when defining agent tools.

Consider:

```python
def add(a: float, b: float) -> float:
```

There are three important pieces.

### First Input

```python
a: float
```

This tells the framework that `a` should be a floating-point number.

### Second Input

```python
b: float
```

This tells the framework that `b` should also be a floating-point number.

### Return Type

```python
-> float
```

This tells the framework that the function returns a floating-point number.

Conceptually:

```text
add(
    a: float,
    b: float
)
→ float
```

These type hints help the framework understand the structure of the tool.

# 15. Docstrings

The function's docstring is also important.

For example:

```python
def add(a: float, b: float) -> float:
    """
    Add two numbers together.
    """
    return a + b
```

The docstring describes what the tool does.

The LLM can use this description to determine when the tool should be selected.

For example:

```text
Tool Name:
add

Description:
Add two numbers together.
```

If the user asks:

```text
What is 42 + 58?
```

The model can recognize that the `add` tool is appropriate.

Therefore:

> A clear tool description can improve tool selection accuracy.

# 16. Why Tool Definitions Matter

Tool definitions are not only documentation for developers.

They also provide structured information that helps the LLM understand:

* What the tool does.
* When to use it.
* What arguments it requires.
* What type of arguments it expects.
* What type of result it returns.

The important components are:

```text
Tool Name
     +
Description
     +
Input Parameters
     +
Input Types
     +
Return Type
```

Together, these help the model make better tool-selection decisions.

---

# 17. From Python Function to JSON Schema

The LLM does not directly receive your Python source code.

For example, the LLM does not see:

```python
def add(a: float, b: float) -> float:
    return a + b
```

Instead, the framework converts the tool definition into a structured representation, commonly a **JSON Schema-like tool definition**.

Conceptually:

```json
{
  "name": "add",
  "description": "Add two numbers together.",
  "parameters": {
    "type": "object",
    "properties": {
      "a": {
        "type": "number"
      },
      "b": {
        "type": "number"
      }
    },
    "required": ["a", "b"]
  }
}
```

The framework sends this structured description to the LLM.

The conceptual flow is:

```text
Python Function
      ↓
Tool Decorator
      ↓
Type Hints + Docstring
      ↓
Tool Schema
      ↓
LLM
```

The LLM sees the structured description and decides whether to call the tool.

---

# 18. Why Type Hints and Docstrings Matter

Type hints and docstrings are especially important in agent applications.

For humans, they provide documentation.

For agents, they can also act as instructions.

Consider:

```python
@tool
def square_root(a: float) -> float:
    """
    Calculate the square root of a number.
    """
    return math.sqrt(a)
```

The LLM can understand:

```text
Tool:
square_root

Use when:
The user needs the square root of a number.

Input:
One number.

Output:
A number.
```

Therefore:

> Tool descriptions, type hints, and function names are important parts of the interface between your code and the LLM.

# 19. How Does the LLM Choose a Tool?

Suppose the agent has four tools:

```text
add
multiply
divide
square_root
```

The user asks:

```text
What is 42 + 58?
```

The LLM looks at the available tools and identifies:

```text
Tool:
add

Description:
Add two numbers together.
```

It then generates a structured tool call.

Conceptually:

```json
{
  "tool": "add",
  "arguments": {
    "a": 42,
    "b": 58
  }
}
```

The framework executes the function:

```python
add(42, 58)
```

The result is:

```text
100
```

The result is then returned to the agent.

The agent can finally respond:

```text
42 + 58 = 100
```

# 24. Example 2 — Multiplication and Division

Now consider:

The agent needs more than one operation.

The correct sequence is:

```text
15 × 8 = 120
120 ÷ 3 = 40
```

The important point is that the developer did not explicitly program this sequence.

The agent dynamically determines the sequence.

---

## Loop 1

The agent reasons:

```text
I need to calculate 15 × 8 first.
```

It selects:

```text
multiply
```

The tool call is:

```text
multiply(
    a=15,
    b=8
)
```

The result is:

```text
120
```

---

## Loop 2

The agent observes the result:

```text
120
```

It then reasons:

```text
Now I need to divide 120 by 3.
```

It selects:

```text
divide
```

The tool call is:

```text
divide(
    a=120,
    b=3
)
```

The result is:

```text
40
```

The agent now has enough information.

Final answer:

```text
15 × 8 ÷ 3 = 40
```

The complete flow is:

```text
User Question
      ↓
Reason
      ↓
Multiply(15, 8)
      ↓
120
      ↓
Reason Again
      ↓
Divide(120, 3)
      ↓
40
      ↓
Final Answer
```

# 25. Tool Result Chaining

One important concept in this example is **result chaining**.

The output of the first tool call becomes the input for the second tool call.

```text
Loop 1:
multiply(15, 8)
      ↓
Result = 120
      ↓
Loop 2:
divide(120, 3)
      ↓
Result = 40
```

The agent dynamically connects the results.

The developer did not explicitly write:

```python
result1 = multiply(15, 8)
result2 = divide(result1, 3)
```

Instead, the agent determines the required sequence based on the user's request.

This is a major characteristic of agentic systems.

---

# 26. Why This Is Different from a Fixed Workflow

Consider a traditional workflow:

```python
result1 = multiply(15, 8)
result2 = divide(result1, 3)
```

The order is explicitly programmed.

The developer decides:

```text
First → Multiply
Second → Divide
```

This is a fixed workflow.

An agent is different.

The agent receives:

```text
15 × 8 ÷ 3
```

It dynamically reasons:

```text
First, I need multiplication.

Then, I need division.
```

The agent decides the execution sequence.

This can be represented as:

### Fixed Workflow

```text
Input
 ↓
Step 1
 ↓
Step 2
 ↓
Step 3
 ↓
Output
```

### Agent

```text
Input
 ↓
Reason
 ↓
Choose Action
 ↓
Observe Result
 ↓
Reason Again
 ↓
Choose Next Action
 ↓
Output
```

# 31. Fixed Workflow vs. Agent

## Fixed Workflow

A fixed workflow might look like:

```text
Input
 ↓
Calculate Area
 ↓
Calculate Square Root
 ↓
Return Result
```

The sequence is predefined.

## Agent

An agent might work like:

```text
User Request
      ↓
Understand Goal
      ↓
Reason
      ↓
Select Tool
      ↓
Execute
      ↓
Observe
      ↓
Reason Again
      ↓
Select Next Tool
      ↓
Execute
      ↓
Observe
      ↓
Final Answer
```

The sequence is dynamically determined.

The key difference is:

> **A workflow follows a predefined sequence, while an agent dynamically decides what actions to take based on the goal and observations.**

# 33. Complete Simplified Python Example

The following example demonstrates the basic concepts using Python-style code.

```python
import math


def add(a: float, b: float) -> float:
    """
    Add two numbers together.
    """
    return a + b


def multiply(a: float, b: float) -> float:
    """
    Multiply two numbers together.
    """
    return a * b


def divide(a: float, b: float) -> float:
    """
    Divide the first number by the second number.
    """
    if b == 0:
        raise ValueError("Cannot divide by zero.")

    return a / b


def square_root(a: float) -> float:
    """
    Calculate the square root of a number.
    """
    if a < 0:
        raise ValueError(
            "Cannot calculate the square root of a negative number."
        )

    return math.sqrt(a)


tools = [
    add,
    multiply,
    divide,
    square_root
]
```

In a real framework, these functions would typically be registered as tools using the framework's tool mechanism.

For example, conceptually:

```python
@tool
def add(a: float, b: float) -> float:
    """
    Add two numbers together.
    """
    return a + b
```

Then the tools can be provided to the agent.

Conceptually:

```python
agent = create_agent(
    model=model,
    tools=tools
)
```

Finally, the agent can be invoked:

```python
response = agent.invoke(
    {
        "input": "What is the square root of the area of a rectangle with width 12 and height 7?"
    }
)
```

The agent may dynamically execute:

```text
multiply(12, 7)
       ↓
84
       ↓
square_root(84)
       ↓
9.165...
```

# 34. Simplified ReAct Agent Pseudocode

The agent's behavior can be represented using pseudocode:

```python
while True:

    # Reason about the user's goal
    decision = llm(
        user_request,
        available_tools
    )

    # Check whether a tool is required
    if decision.requires_tool:

        # Execute the selected tool
        result = execute_tool(
            decision.tool,
            decision.arguments
        )

        # Add the result to the agent's context
        context.append(result)

        # Continue reasoning
        continue

    # If no tool is required,
    # return the final answer
    return decision.final_answer
```

The key pattern is:

```text
Reason
  ↓
Choose Tool
  ↓
Act
  ↓
Observe
  ↓
Reason Again
  ↓
Stop or Continue
```

### Q7. What is the difference between a workflow and an agent?

A workflow generally follows a predefined sequence of steps. An agent dynamically decides which actions to take and what tools to use based on the user's goal and the results it observes.

---

### Q8. What happens when an agent needs multiple tools?

**Answer:**

The agent can execute tools iteratively. The result of one tool call can be used as input for a subsequent tool call.

Example:

```text
multiply(12, 7)
       ↓
84
       ↓
square_root(84)
       ↓
9.165...
```

The core formula is:

```text
LLM
+
Tools
+
Agent Loop
=
AI Agent
```

```text
1. Choose the LLM
        ↓
2. Define useful tools
        ↓
3. Create the agent
        ↓
4. Invoke the agent
        ↓
5. Agent reasons about the task
        ↓
6. Agent selects and calls tools
        ↓
7. Agent observes results
        ↓
8. Agent continues or stops
        ↓
9. Final response
```
