# Introduction to LangChain

## Overview

LangChain is an **open-source framework for building applications powered by Large Language Models (LLMs)**.

An LLM by itself is powerful. It can:

- Generate text
- Answer questions
- Summarize information
- Reason about problems
- Follow instructions

However, an LLM by itself has important limitations.

For example, a standalone LLM generally cannot directly:

- Call external APIs
- Query databases
- Execute application code
- Access external tools
- Interact with real-world systems

LangChain helps solve this problem by connecting LLMs with external systems and application logic.

The key idea is:

```text
LLM
  ↓
Connect to external systems
  ↓
Orchestrate multiple steps
  ↓
Build intelligent applications
