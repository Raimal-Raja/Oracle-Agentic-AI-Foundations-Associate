# Module Introduction: Model Context Protocol (MCP)

Welcome to the module on **Introduction to MCP**.

MCP stands for **Model Context Protocol**. It is an open standard introduced by Anthropic in November 2024 that standardizes how AI models connect with external tools, applications, and data sources.

In this module, we will explore MCP through four focused lessons.

### Lesson 1: Understanding MCP

First, we will build a strong foundation by understanding **what MCP is and why it exists**.

We will look at the problems MCP solves and why a standardized approach is important when connecting AI models with external systems.

### Lesson 2: How MCP Works

Next, we will break down the MCP architecture and understand how the protocol actually works.

We will explore the core components of MCP, including:

* **Hosts**
* **Clients**
* **Servers**
* **Primitives**

Understanding these components will help us see how AI applications communicate with external tools and data through MCP.

### Lesson 3: MCP in an AI Agent

In the third lesson, we will take a working AI agent and see how MCP changes its architecture in practice.

We will use the same simple math agent that we introduced in the earlier modules. If you remember, that agent uses a simple math question that requires multiple tool calls.

We are intentionally using the same example from **Modules 1 and 2** to maintain continuity throughout the course.

This time, however, we will create a new **MCP Math Server** and compare the architecture before and after introducing MCP.

We will look at:

**Before MCP**

Your agent directly connects to and manages its tools.

**After MCP**

The agent communicates with tools through an MCP server.

By comparing these two architectures, you will clearly understand how MCP changes the way tools are connected to AI agents.

### Lesson 4: A Real-World MCP Example

Finally, we will move from a learning example to a real-world scenario.

The MCP server we create in this module is primarily designed for learning and demonstration purposes. In real production environments, MCP servers are often created and published by other teams, companies, or service providers.

In those situations, you typically do not build the MCP server yourself. Instead, your main responsibility is to build and configure the **MCP client** that connects your AI application or agent to the existing server.

We will therefore look at what changes in your application code when you move from building your own MCP server to consuming a production-ready MCP server.

### The Goal of This Module

The goal of this module is simple.

By the end of this module, you should not only understand **MCP conceptually**, but also understand **exactly where MCP fits into real-world AI agent architectures**.

You will understand:

* What MCP is and why it was created.
* How MCP works internally.
* The roles of hosts, clients, servers, and primitives.
* How MCP changes the architecture of an AI agent.
* How an agent communicates with tools through MCP.
* The difference between building an MCP server and consuming an existing one.
* How MCP can be applied in real-world production systems.
