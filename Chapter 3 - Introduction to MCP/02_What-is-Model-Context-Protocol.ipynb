# Introduction to MCP

Welcome to the module on **Introduction to MCP**.

MCP stands for **Model Context Protocol**. It is an open standard introduced by Anthropic in November 2024 that standardizes how AI models connect to external tools and data sources.

In this module, we will explore MCP through four focused lessons.

## Lesson 1: What Is MCP and Why Does It Exist?

First, we will build a strong foundation by understanding **what MCP is and why it was created**.

We will explore the problems MCP solves and why having a standardized protocol is important for connecting AI models with external tools, services, and data sources.

The goal is to understand the basic concept of MCP before diving into its technical architecture.

## Lesson 2: How MCP Works

Next, we will look at how MCP actually works.

We will explore the core components of the MCP architecture, including:

* **Hosts**
* **Clients**
* **Servers**
* **Primitives**

Understanding these components will help us see how AI applications communicate with external tools and data sources using MCP.

## Lesson 3: MCP and AI Agents

In the third lesson, we will take a working AI agent and see how MCP changes its architecture in practice.

We will use the same simple agent example that we introduced in **Modules 1 and 2**.

If you remember, the agent uses a simple math problem that requires multiple tool calls. We are intentionally using the same example to maintain continuity throughout the course.

This time, we will create a new **MCP Math Server** and compare the architecture before and after introducing MCP.

We will examine:

**Before MCP**

The AI agent directly connects to and manages its tools.

**After MCP**

The AI agent communicates with external tools through an MCP server.

By comparing these two architectures, you will clearly understand how MCP changes both the architecture and the code of an AI agent.

## Lesson 4: MCP in the Real World

Finally, we will look at a real-world example of MCP.

The MCP server we create in this module is primarily designed for learning and demonstration purposes. In real-world production environments, MCP servers are typically created and published by other developers, teams, or organizations.

In that situation, you generally do not build the MCP server yourself. Instead, you write an **MCP client** that connects your AI application or agent to an existing MCP server.

We will explore what changes in your code when you move from building your own MCP server to consuming an MCP server that has already been published.

We will also identify which parts of your application remain the same and which parts need to change.

## The Goal of This Module

The goal of this module is simple.

By the end of this module, you should not only understand **MCP conceptually**, but also understand **exactly where MCP fits into real-world AI systems and agent architectures**.

You will understand:

* What MCP is.
* Why MCP was created.
* How MCP works.

* How MCP changes the architecture of an AI agent.
