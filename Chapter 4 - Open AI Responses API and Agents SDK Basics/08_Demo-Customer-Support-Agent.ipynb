# Project Overview

The project implements an AI-powered customer support system consisting of one routing agent and multiple specialized agents.

Instead of one AI handling every request, the Triage Agent determines which specialist is best suited to answer the customer's question.

This architecture follows the principle of **specialized AI agents working collaboratively**.

---

# System Architecture

```
                    Customer
                        │
                        ▼
              Input Guardrail
                        │
                        ▼
                 Triage Agent
        ┌──────────┼───────────┐
        ▼          ▼           ▼
 Order Status   Refund      FAQ Agent
     Agent       Agent
        │          │
        ▼          ▼
 Custom Tool   Custom Tool
        │
        ▼
 Python Function / Database

FAQ Agent
        │
        ▼
 OpenAI Web Search Tool
```

---

# Components of the System

## 1. Input Guardrail

The Input Guardrail validates incoming customer messages before they reach the Triage Agent.

Its responsibilities include:

- Blocking off-topic questions
- Preventing malicious prompts
- Rejecting unsupported requests
- Ensuring only customer support requests continue

If validation fails, the request is immediately stopped.

---

## 2. Triage Agent

The Triage Agent acts as the central coordinator.

Its responsibilities include:

- Receiving customer requests
- Understanding customer intent
- Selecting the appropriate specialist
- Performing handoffs

The Triage Agent does not solve customer problems directly.

Instead, it delegates work to specialized agents.

---

## 3. Order Status Agent

The Order Status Agent is responsible for handling order-related inquiries.

Examples include:

- Where is my order?
- Order tracking
- Shipping status
- Delivery information

This agent uses a **custom Python function tool** to retrieve order information.

---

## 4. Refund Agent

The Refund Agent processes refund-related requests.

Typical tasks include:

- Refund eligibility
- Refund status
- Refund processing

Like the Order Status Agent, it also uses custom Python functions.

---

## 5. FAQ Agent

The FAQ Agent answers general customer questions.

Unlike the previous agents, it does not require custom Python code.

Instead, it uses the OpenAI built-in Web Search Tool.

Examples include:

- Company policies
- Business hours
- General product information
- Frequently asked questions

# Multi-Agent Collaboration

Each agent has a clearly defined responsibility.

```
Customer Question
        │
        ▼
Input Guardrail
        │
        ▼
Triage Agent
        │
        ├────────► Order Status Agent
        │
        ├────────► Refund Agent
        │
        └────────► FAQ Agent
```

This design follows the **Single Responsibility Principle**, making the system easier to extend and maintain.

# Customer Request Flow

Consider the following customer request:

```
Where is my order?

Order ID:
ORD-001
```

The complete execution flow is:

### Step 1

Customer submits the request.

↓

### Step 2

Input Guardrail validates the request.

Question:

Is this a customer support request?

↓

### Step 3

Validation succeeds.

↓

### Step 4

The Triage Agent analyzes the request.

↓

### Step 5

The Triage Agent determines that this is an order-related query.

↓

### Step 6

A handoff transfers control to the Order Status Agent.

↓

### Step 7

The Order Status Agent calls the Lookup Order Tool.

↓

### Step 8

The Python function retrieves order details.

↓

### Step 9

The returned data is sent back to the LLM.

↓

### Step 10

The LLM generates a natural language response.

↓

### Step 11

The customer receives the final answer.

# Understanding the Handoff

A handoff transfers execution from one AI agent to another.

Example:

```
Customer

↓

Triage Agent

↓

Order Status Agent
```

Important:

The handoff itself is extremely fast because it simply changes which agent is currently responsible.

No additional LLM reasoning is required for the handoff mechanism itself.

# Custom Function Tools

Two specialized agents use custom Python functions.

Examples include:

Order Lookup

```
lookup_order(order_id)
```

Refund Processing

```
process_refund(order_id)
```

These functions execute locally or remotely and return structured data to the AI model.

# Hosted Tools

The FAQ Agent uses the OpenAI Hosted Web Search Tool.

Unlike custom functions:

- No Python implementation is required.
- The SDK manages execution.
- Internet information can be retrieved automatically.

# Tracing

Tracing automatically records every step performed by the AI system.

It provides complete visibility into:

- Guardrail execution
- Agent decisions
- Handoffs
- Tool calls
- Model responses
- Token usage
- Execution time

Tracing is one of the most valuable debugging and monitoring features of the OpenAI Agents SDK.

# Phase 1 – Input Guardrail

The first component to execute is **not** the Triage Agent.

Instead:

```
Customer

↓

Support Topic Checker

↓

LLM Validation

↓

Pass / Fail
```

If the request is not related to customer support, execution stops immediately.

Example:

```
What's the weather today?
```

↓

Blocked.

Example:

```
Where is my order?
```

↓

Allowed.

# Phase 2 – Triage Agent

After validation:

The Triage Agent analyzes customer intent.

Example:

```
Where is my order?
```

↓

Recognized as

```
Order Status Request
```

↓

Handoff occurs.

# Why Tool Calls Are Fast

The lesson highlights an important observation.

The custom Python functions execute in only a few milliseconds.

Examples include:

- Database lookup
- Dictionary lookup
- Local computation

These operations are significantly faster than LLM API requests.

---

# Performance Analysis

Example execution:

| Component | Approximate Time |
|------------|-----------------|
| Guardrail Validation | ~5 seconds |
| Triage Agent | ~2 seconds |
| Order Status Agent | ~2 seconds |
| Lookup Tool | ~2 milliseconds |
| Response Generation | ~1.5 seconds |

The primary bottleneck is the LLM API, not the Python code.

---

# Production Insights

Tracing helps identify optimization opportunities.

For example:

Instead of optimizing Python functions that execute in milliseconds, developers should focus on reducing:

- Number of LLM calls
- Prompt size
- Context length
- Token consumption

This results in significant performance improvements.

# Best Practices

- Keep agents specialized.
- Use handoffs instead of large monolithic prompts.
- Protect systems with Input Guardrails.
- Prefer hosted tools when appropriate.
- Use custom tools for business logic.
- Monitor traces regularly.
- Optimize LLM usage before optimizing Python code.
- Measure token consumption continuously.

# Interview Questions

Multiple agents improve modularity, scalability, maintainability, and specialization by assigning different tasks to dedicated AI agents.

---

## What is a Triage Agent?

A Triage Agent is responsible for analyzing incoming requests and routing them to the appropriate specialized agent.

---

## What is a handoff?

A handoff transfers execution from one AI agent to another without requiring the entire workflow to restart.

---

## Why are custom tools useful?

Custom tools allow AI agents to interact with external systems such as databases, APIs, business logic, and enterprise applications.
