# OCI Enterprise AI Platform

## Overview

Oracle Cloud Infrastructure (OCI) Enterprise AI Platform is Oracle's comprehensive platform for building, deploying, managing, and governing enterprise-grade AI applications. Rather than providing only AI models, the platform delivers a complete ecosystem that enables organizations to create secure, scalable, and production-ready AI solutions.

The platform is organized into three major layers:

- Enterprise AI Models (Intelligence Layer)
- Enterprise AI Agents Service (Action Layer)
- Enterprise AI Governance (Trust Layer)

Together, these layers provide AI intelligence, execution capabilities, and enterprise-level security.

# Types of Enterprise AI Models

OCI primarily supports three categories of AI models.

- Chat Models
- Embedding Models
- Rerank Models

Each serves a different purpose.

---

# Chat Models

Chat models are the most commonly used AI models.

They generate natural language responses based on user prompts.

Typical capabilities include:

- Question answering
- Text generation
- Summarization
- Code generation
- Multi-turn conversations
- Tool calling
- AI agents

---

## Characteristics

- Stateless by default
- Accept prompts or conversations
- Generate natural language responses
- Support tool calling
- Can participate in agent workflows

---

## Conversation Memory

By default, chat models do not remember previous requests.

Conversation history is maintained using the **OCI Conversations API**, which enables multi-turn interactions.

---

## Temperature

Temperature controls randomness during response generation.

### Low Temperature

- More deterministic
- More factual
- Consistent outputs

Example:

```
Temperature = 0.1
```

---

### High Temperature

- More creative
- More varied
- Less predictable

Example:

```
Temperature = 0.9
```

---

# Embedding Models

Embedding models do not generate text.

Instead, they convert text (or other supported data types) into numerical vectors.

Example:

```
Input

"Artificial Intelligence"

↓

Embedding Model

↓

[0.123, -0.456, 0.789, ...]
```

These vectors capture semantic meaning.

---

# Semantic Similarity

One of the most important properties of embeddings is semantic similarity.

Sentences with similar meanings produce vectors located close together in vector space.

Example:

Sentence A

```
I purchased a laptop.
```

Sentence B

```
I bought a notebook computer.
```

Although the wording differs, the embedding vectors are very similar.

---

# Semantic Search

Embedding models enable semantic search.

Workflow:

```
User Query

↓

Embedding Model

↓

Vector

↓

Vector Database

↓

Nearest Matching Documents
```

Unlike keyword search, semantic search retrieves documents based on meaning rather than exact words.

---

# Multimodal Embeddings

Some embedding models support multiple data types.

Example:

**Cohere Embed 4**

Supports:

- Text
- Images

Both are converted into a shared vector space.

This enables cross-modal search.

Example:

Image → Similar Text

or

Text → Similar Images

---

# Rerank Models

Rerank models improve document retrieval quality in Retrieval-Augmented Generation (RAG) systems.

Unlike embedding models, rerank models do not retrieve documents.

Instead, they reorder previously retrieved documents according to relevance.

---

# What is RAG?

RAG stands for:

**Retrieval-Augmented Generation**

RAG combines external knowledge with LLM reasoning.

Workflow:

```
User Question

↓

Embedding Search

↓

Candidate Documents

↓

Rerank Model

↓

Most Relevant Documents

↓

Chat Model

↓

Final Answer
```

---

# Why Reranking Matters

Suppose semantic search retrieves 50 candidate documents.

Many of these documents may only be partially relevant.

A rerank model scores each document according to the user's query.

Example:

Before Reranking

```
50 Documents
```

↓

After Reranking

```
Top 5 Documents
```

Only the highest-quality context is sent to the chat model.

This significantly improves answer quality.

---

# RAG Quality Chain

Successful RAG systems depend on three different model types.

| Model | Responsibility |
|--------|----------------|
| Embedding Model | Recall (Find Relevant Documents) |
| Rerank Model | Precision (Select Best Documents) |
| Chat Model | Response Generation |

All three components work together to produce accurate answers.

# OCI Model Serving Modes

OCI provides two methods for serving AI models.

- On-Demand
- Dedicated AI Clusters

Choosing the correct serving mode affects:

- Cost
- Performance
- Scalability
- Reliability
- Compliance

---

# On-Demand Serving

On-Demand is designed for experimentation and development.

Characteristics:

- Pay per API request
- No upfront commitment
- Shared GPU infrastructure
- Quick setup
- Ideal for prototypes

---

## Advantages

- Low initial cost
- Easy to start
- Suitable for testing
- Flexible pricing

---

## Limitations

- Shared infrastructure
- Dynamic throttling
- Rate limiting
- Model retirement may interrupt access
- No support for custom models

---

# Dedicated AI Clusters (DAC)

Dedicated AI Clusters provide enterprise-grade production infrastructure.

Characteristics:

- Dedicated GPUs
- Predictable latency
- High availability
- Supports custom models
- Supports imported models
- Better isolation

---

## Advantages

- Stable performance
- SLA support
- GPU isolation
- Production-ready
- Continued access after model retirement
- Better compliance

---

## Cost Model

Instead of paying per request, organizations pay for cluster uptime.

Billing is based on active cluster hours.

---

# On-Demand vs Dedicated AI Clusters

| Feature | On-Demand | Dedicated AI Cluster |
|----------|-----------|----------------------|
| Cost Model | Pay per request | Pay per cluster hour |
| GPU | Shared | Dedicated |
| Startup | Immediate | Provision Required |
| Scalability | Shared | Dedicated |
| Latency | Variable | Predictable |
| Custom Models | No | Yes |
| Production Ready | Limited | Yes |

---

# Choosing the Right Serving Mode

Use **On-Demand** when:

- Learning
- Experimenting
- Building prototypes
- Validating ideas

Use **Dedicated AI Clusters** when:

- Deploying production systems
- Requiring predictable latency
- Using custom models
- Meeting compliance requirements
- Serving enterprise workloads

# Enterprise AI Governance

Enterprise AI Governance provides security, compliance, and control across the entire AI platform.

Think of this as the **Trust Layer**.

Its purpose is to ensure AI systems operate securely and responsibly.

---

# Three Layers of Governance

OCI Enterprise AI Governance is built upon three security layers.

- Network Security
- Identity Control
- AI Behavior Control

---

# 1. Network Security

Network Security controls how requests reach AI services.

Important technologies include:

- Private Endpoints
- Zero Trust Packet Routing

---

## Private Endpoints

Private endpoints keep AI traffic inside secure OCI networks.

Benefits include:

- Reduced public exposure
- Improved security
- Private communication

---

## Zero Trust Packet Routing

Zero Trust assumes no network connection should be automatically trusted.

Every communication must be authenticated and authorized.

---

# 2. Identity Control

Identity Control determines who can access AI resources.

OCI provides this through:

**Identity and Access Management (IAM)**

IAM controls:

- Users
- Groups
- Roles
- Permissions
- Policies

Administrators can restrict access to individual AI resources or entire OCI compartments.

---

# 3. AI Behavior Control

Even authenticated users may submit unsafe prompts.

Behavior Control protects the AI model during runtime.

Key capabilities include:

- Prompt Injection Detection
- Input Moderation
- Output Moderation
- Content Filtering
- Personally Identifiable Information (PII) Detection
- Guardrails

These controls ensure safe AI behavior.

---

# Defense-in-Depth

OCI follows a Defense-in-Depth security strategy.

Multiple security layers work together.

```
Network Security

↓

Identity & Access Management

↓

Runtime Guardrails

↓

AI Model
```

Each layer protects against different threats.

---
